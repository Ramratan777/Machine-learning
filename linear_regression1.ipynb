Model function

f_wb = w*x[0] + b

```python
w=1
b=2
print(f"w: {w}")
print(f"b: {b}")
```

```python
def predict(x, w, b):
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

```python
tmp_f_wb = predict(x_train, w, b)
print(f"tmp_f_wb: {tmp_f_wb}")
plt.scatter(x_train, y_train, color='blue', label='Training Data')
plt.plot(x_train, tmp_f_wb, color='red', label='Prediction')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Training Data and Prediction')
plt.legend()
plt.grid()
plt.show()  
```

# cost function

```python
np.random.seed(42)
x_train=np.random.randint(1,101,10)
print(f"x_train: {x_train}")
```

```python

y_train=np.random.randint(1,101,10)
print(f"y_train: {y_train}")
```

```python

# Cost function
def compute_cost(w, b):
    m = len(x_train)
    predictions = w * x_train + b
    return np.sum((predictions - y_train) ** 2) / (2 * m)

# Create grid of w and b values
w_vals = np.linspace(-4, 4, 100)
b_vals = np.linspace(-4, 4, 100)

W, B = np.meshgrid(w_vals, b_vals)

# Compute cost for each (w,b)
J = np.zeros_like(W)

for i in range(W.shape[0]):
    for j in range(W.shape[1]):
        J[i, j] = compute_cost(W[i, j], B[i, j])

# 3D plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(W, B, J, cmap='viridis')

ax.set_xlabel('w')
ax.set_ylabel('b')
ax.set_zlabel('Cost J(w,b)')
ax.set_title('Cost Function Surface')

plt.show()
```

```python
row, col = np.unravel_index(np.argmin(J), J.shape)

print(row, col)
```

```python
plt.scatter(x_train, y_train)
plt.xlabel("x_train")
plt.ylabel("y_train")
plt.title("Training Data")
plt.grid()
plt.show()
```

```python
predict_function = predict(x_train, W[row, col], B[row, col])
print(f"predict_function: {predict_function}")
```

```python
plt.scatter(x_train, y_train, color='blue', label='Training Data')
plt.plot(x_train, predict_function, color='red', label='Prediction')
plt.xlabel("x_train")
plt.ylabel("y_train")
plt.title("Training Data and Prediction")
plt.grid()
plt.show()
```
